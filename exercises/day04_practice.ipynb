```python
import numpy as np
import pandas as pd

bees = pd.Series([True, True, False, np.nan, True, False, True, np.nan])
knees = pd.Series([5,2,9,1,3,10,5,2], index = [7,0,2,6,3,5,1,4])

print(bees)
print(knees)
```

```python
knees.iloc[3] = knees.iloc[3] * 2
knees
```

```python
# Given, two Series bees and knees, if the ith value of bees is NaN, double the ith value inside knees.
for x,y in bees.items():
    if pd.isna(y):
        knees.iloc[x] = knees.iloc[x] * 2


knees
```

```python
asking_prices = pd.Series([5000, 7600, 9000, 8500, 7000], index=['civic', 'civic', 'camry', 'mustang', 'mustang'])
fair_prices = pd.Series([5500, 7500, 7500], index=['civic', 'mustang', 'camry'])

print(asking_prices)
print(fair_prices)
```

```python
# Given a Series of car asking_prices and another Series of car fair_prices, determine which cars for sale are a good deal. In other words, identify cars whose asking price is less than their fair price.
for idx in fair_prices.index:
    # if 
    for val in asking_prices[[idx]].values:
        if val < fair_prices[idx]:
            print(fair_prices[idx], val)
            
```

```python
df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Age": [20, 25, 30, 22],
    "Salary": [30000, 50000, 70000, 40000]
})
df[df["Salary"] > 40000]
df[(df["Salary"] > 40000) & (df["Age"] > 25)]
```

```python
df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [50000, 40000, 60000, 45000, 70000]
})
df.groupby("Department")["Salary"].mean()
df.groupby("Department")["Salary"].agg(["mean", "max"])
```

```python
df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Age": [20, np.nan, 25, np.nan],
    "Salary": [30000, 40000, np.nan, 50000]
})

# df.dropna()
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df["Salary"].fillna(df['Salary'].mean())
df
```
