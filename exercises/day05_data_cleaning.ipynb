```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": [
        "Rahul", "Priya", "Amit", "rahul ", "Neha",
        "Vikas", "Priya", "Rohan", "Sneha", "Amit"
    ],
    
    "Age": [
        22, np.nan, 25, 22, "twenty-four",
        -5, np.nan, 30, 150, 25
    ],
    
    "City": [
        "Lucknow", "lucknow", "Delhi", "Lucknow ",
        "DELHI", "Kanpur", "Lucknow", "Delhi", "delhi", "Delhi"
    ],
    
    "Gender": [
        "Male", "Female", "M", "male", "F",
        "Female", "female", "Male", "F", "Male"
    ],
    
    "Salary": [
        35000, 42000, np.nan, 35000, 50000,
        38000, 42000, "45000", 48000, 45000
    ],
    
    "Joining_Date": [
        "2022-01-15", "15/02/2021", "2020-03-10", "2022-01-15",
        "invalid_date", "2023/05/20", "15-02-2021", "2021-12-01",
        "2020-13-40", "2020-03-10"
    ]
}

df = pd.DataFrame(data)

df
```

## clean invalid value

```python
df.info()
```

```python
df.describe()
df.isnull().sum()
```

```python
df['Age'].value_counts(dropna=False)

# df[df['Age'] < 0]
# df[df['Age'] > 100]
```

```python
invalid_row = df[df['Age'] == 'twenty-four']
idx = invalid_row.index.start
df.loc[idx, 'Age'] = 24
print(df['Age'].dtypes)
df['Age']
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
print(df['Age'].dtypes)
```

```python
df.loc[df['Age'] < 0, 'Age'] = np.nan

df.loc[df['Age'] > 100, 'Age'] = np.nan
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Age'].isnull().sum()
```

```python
df.isnull().sum()
```

## String Cleaning

```python
df['City'].value_counts()
df['City'] = df['City'].str.strip()
df['City'] = df['City'].str.lower()

df['City'].value_counts()
df["City"].unique()
```

```python
df["Gender"].unique() # ['Male', 'Female', 'M', 'male', 'F', 'female']
df["Gender"].value_counts()
```

```python
df['Gender'] = df['Gender'].replace({
    'M': 'Male',
    'male': 'Male',
    'female': 'Female',
    'F': 'Female'
})
df['Gender'].value_counts()
```

```python
df["Name"].unique()

df['Name'] =  df['Name'].str.strip().str.title()

df['Name'].value_counts()
```

## Date / Time Cleaning

```python
df['Joining_Date']
```

```python
print(df["Joining_Date"].dtype) # str

df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    errors="coerce"
)


df['Joining_Date']
```

```python
print(df["Joining_Date"].dtype)
df["Joining_Date"].dt.year
df["Joining_Date"].dt.month
```

```python
dates = pd.Series(["15/02/2021", "20/03/2022", "05/11/2023"])
dates = pd.to_datetime(dates, format='%d/%m/%Y')
print(dates.dtypes)
```

```python
df['Joining_Date'] = pd.to_datetime(
    df['Joining_Date'],
    format="mixed",
    errors="coerce"
)
df
```

```python
df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
df
```
