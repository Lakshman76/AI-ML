## Series

```python
import numpy as np
import pandas as pd
```

### Series from list

```python
countries = ['India', 'Nepal', 'Bhutan', 'srilanka', 'USA']
pd.Series(countries)
```

```python
runs = [30, 48, 23, 56, 67, 56, 87, 34]
run_series = pd.Series(runs)
run_series
```

```python
marks = [89, 90, 92, 95, 100]
subject = ['Hindi', 'English', 'SST', 'Science', 'Maths']
pd.Series(marks, index=subject)
```

```python
marks = pd.Series(marks, index=subject, name='Laksh ke marks')
marks
```

### Series from Dictionary

```python
marks = {
    'math': 99,
    'science': 95,
    'sst': 88,
    'eng': 91,
    'Hindi': 75,
    'computer': 100
}

marks_series = pd.Series(marks, name='Laksh2 ke marks')
marks_series
```

```python
# drop_duplicates
temp = pd.Series([1, 1, 2, 2, 2, 3, 3, 4, 4, 5])
temp.drop_duplicates()
```

```python
temp.drop_duplicates(keep='last')
```

```python
temp[temp.duplicated()]
temp.duplicated().sum()
```

```python
movies.size
movies.drop_duplicates().size
```

```python
# isnull
temp = pd.Series([1,2,3,np.nan,5,6,np.nan,8,np.nan,10])
temp.size # 10
temp.count() # 7
```

```python
temp.isnull()
temp.isnull().sum()
```

```python
# dropna
temp.dropna()
```

```python
# fillna
temp.fillna(0)
temp.fillna(temp.mean())
```

```python
# isin
vk_run[(vk_run == 49) | (vk_run == 99) | (vk_run == 50)]
vk_run[vk_run.isin([49, 99, 50 , 100])]
```
