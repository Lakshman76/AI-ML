# Day 01 - Pandas Basics

Learn the fundamentals of Pandas: Series, DataFrames, inspection, selection, statistics, and creating new columns.

```python
import numpy as np
import pandas as pd

print('Pandas Version:', pd.__version__)
```

## 1. Creating a Series

```python
series = pd.Series([10, 20, 30, 40, 50])
series
```

```python
custom_series = pd.Series([100, 200, 300], index=['A', 'B', 'C'])
custom_series
```

## 2. Creating a DataFrame

```python
data = {
    'Name': ['Lakshman', 'Rahul', 'Aman', 'Naman', 'Himanshu', 'Bhanu', 'Akash'],
    'Age': [22, 23, 21, 24, 23, 23, 24],
    'City': ['Lucknow', 'Delhi', 'Mumbai', 'Pune', 'Gzb', 'Saharan', 'mirza']
}

df = pd.DataFrame(data)
df
```

## 3. Inspecting Data

```python
df.head()
```

```python
df.tail()
```

```python
print('Shape:', df.shape)
print('Columns:', list(df.columns))
print('Index:', df.index)
```

```python
df.info()
```

## 4. Selecting Columns

```python
df['Name']
```

```python
df[['Name', 'Age']]
```

### Selecting Rows

```python
# Fetch 1st row
df.iloc[0]

# Fetch 1st to 3rd (index 0 - 2)
df.iloc[0:3]

#Fetch from 1st to 20th by skiping 3 row
df.iloc[0:20:3]

# Fetch 1, 3 and 6th row
df.iloc[[1, 3, 6]]

#Fetch all row of 0, 1 and 2nd column (0-based index)
df.iloc[:,[0, 1, 2]]
```

## 5. Data Types

```python
df.dtypes
```

## 6. Basic Statistics

```python
df.describe()
```

```python
print('Average Age:', df['Age'].mean())
print('Maximum Age:', df['Age'].max())
print('Minimum Age:', df['Age'].min())
```
